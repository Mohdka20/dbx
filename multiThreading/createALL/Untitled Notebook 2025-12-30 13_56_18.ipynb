```python
%sql
CREATE TABLE dev_hub_bronze.lh_ax_idr.tcrrhdr (
  lh_actn_utc_dttm TIMESTAMP,
  data_received_utc_dttm TIMESTAMP,
  source_file_path STRING,
  DRV_REQ_NBR INT,
  AX_CLNT_NBR INT,
  SBRN_PGM_ACNT_NBR SMALLINT,
  SO_ID SMALLINT,
  CONV_DRV_REQ_NBR INT,
  CONV_SPLY_REQ_ID VARCHAR(40),
  REQ_DB_LOC_CDE VARCHAR(2),
  REQ_STE_CDE VARCHAR(2),
  REQ_STE_DT DATE,
  AUDT_USER_ID VARCHAR(8),
  AUDT_INS_TSTMP TIMESTAMP,
  AUDT_TSTMP TIMESTAMP,
  AUDT_ACTN_CDE VARCHAR(1),
  AUDT_PGM_ID VARCHAR(15),
  AUD_CDD_ENTTYP VARCHAR(30),
  AUD_CDD_TSTMP TIMESTAMP,
  AUD_CDD_METHOD VARCHAR(30),
  DW_CLOUD_INSR_DTTM TIMESTAMP,
  DW_CLOUD_UNIQID VARCHAR(36),
  SBRN_PGM_ID VARCHAR(1))

TBLPROPERTIES (
  'delta.columnMapping.mode' = 'name',
  'delta.enableDeletionVectors' = 'true',
  'delta.feature.columnMapping' = 'supported',
  'delta.feature.deletionVectors' = 'supported',
  'delta.minReaderVersion' = '3',
  'delta.minWriterVersion' = '7')
```

```python
%sql
INSERT INTO dev_hub_bronze.lh_ax_idr.tcrrhdr
SELECT
  current_timestamp() AS lh_actn_utc_dttm,
  current_timestamp() AS data_received_utc_dttm,
  concat('/path/file_', seq) AS source_file_path,
  seq AS DRV_REQ_NBR,
  seq + 1000 AS AX_CLNT_NBR,
  seq % 100 AS SBRN_PGM_ACNT_NBR,
  seq % 10 AS SO_ID,
  seq + 2000 AS CONV_DRV_REQ_NBR,
  concat('supply_req_', seq) AS CONV_SPLY_REQ_ID,
  lpad(seq, 2, '0') AS REQ_DB_LOC_CDE,
  lpad(seq, 2, '0') AS REQ_STE_CDE,
  date_add(current_date(), seq) AS REQ_STE_DT,
  concat('user', seq) AS AUDT_USER_ID,
  current_timestamp() AS AUDT_INS_TSTMP,
  current_timestamp() AS AUDT_TSTMP,
  'A' AS AUDT_ACTN_CDE,
  concat('pgm', seq) AS AUDT_PGM_ID,
  concat('enttyp', seq) AS AUD_CDD_ENTTYP,
  current_timestamp() AS AUD_CDD_TSTMP,
  concat('method', seq) AS AUD_CDD_METHOD,
  current_timestamp() AS DW_CLOUD_INSR_DTTM,
  uuid() AS DW_CLOUD_UNIQID,
  lpad(seq, 1, '0') AS SBRN_PGM_ID
FROM (SELECT posexplode(array_repeat(1, 100)) AS (seq, val))
```

```python
%sql
INSERT INTO dev_hub_bronze.lh_ax_idr.tcrstat
SELECT
  current_timestamp() AS lh_actn_utc_dttm,
  current_timestamp() AS data_received_utc_dttm,
  concat('/path/file_', seq) AS source_file_path,
  seq AS DRV_REQ_NBR,
  seq + 1000 AS AX_CLNT_NBR,
  lpad(seq, 2, '0') AS SBRN_PGM_ID,
  seq % 100 AS SBRN_PGM_ACNT_NBR,
  seq % 10 AS SO_ID,
  seq % 5 AS REQ_VERS_NBR,
  lpad(seq, 2, '0') AS CLNT_REQ_SCDE,
  current_timestamp() AS AUDT_INS_TSTMP,
  date_add(current_date(), seq) AS STAT_XDT,
  date_add(current_date(), seq + 1) AS STAT_EDT,
  lpad(seq, 2, '0') AS CLNT_REQ_STAT_TCDE,
  lpad(seq, 2, '0') AS REQ_STAT_CAT_CDE,
  lpad(seq, 2, '0') AS REQ_STAT_RCDE,
  lpad(seq, 2, '0') AS OP_CELL_TCDE,
  concat('user', seq) AS AUDT_USER_ID,
  concat('pgm', seq) AS AUDT_PGM_ID,
  current_timestamp() AS AUDT_TSTMP,
  'A' AS AUDT_ACTN_CDE,
  CASE WHEN seq % 2 = 0 THEN 'Y' ELSE 'N' END AS RCVR_IND,
  current_timestamp() AS ORIG_REQ_INS_TSTMP,
  0 AS INS_USER_ID,
  lpad(seq, 4, '0') AS INS_WLOC_CDE,
  concat('enttyp', seq) AS AUD_CDD_ENTTYP,
  current_timestamp() AS AUD_CDD_TSTMP,
  concat('method', seq) AS AUD_CDD_METHOD,
  current_timestamp() AS DW_CLOUD_INSR_DTTM,
  uuid() AS DW_CLOUD_UNIQID,
  concat('elmt', seq) AS ELMT_ID_NBR,
  seq AS RSK_CRIT_ID,
  current_timestamp() AS EVNT_DTTM
FROM (SELECT posexplode(array_repeat(1, 100)) AS (seq, val))
```

```python


# TCRRLCL
df_tcrrlcl = (
    spark.range(0, 10)
    .withColumn("lh_actn_utc_dttm", current_timestamp())
    .withColumn("data_received_utc_dttm", current_timestamp())
    .withColumn("source_file_path", concat(lit("/path/file_"), col("id")))
    .withColumn("DRV_REQ_NBR", when(col("id") < 8, col("id")).otherwise(lit(2)))
    .withColumn("AX_CLNT_NBR", when(col("id") < 8, col("id") + 1000).otherwise(lit(1002)))
    .withColumn("SBRN_PGM_ID", when(col("id") < 8, lpad(col("id"), 2, "0")).otherwise(lit("02")))
    .withColumn("SBRN_PGM_ACNT_NBR", when(col("id") < 8, col("id") % 100).otherwise(lit(2)))
    .withColumn("SO_ID", when(col("id") < 8, col("id") % 10).otherwise(lit(2)))
    .withColumn("REQ_VERS_NBR", when(col("id") < 8, col("id") % 5).otherwise(lit(2)))
    .withColumn("CLNT_FUNC_CDE", when(col("id") < 8, lpad(col("id"), 2, "0")).otherwise(lit("02")))
    .withColumn("RLTD_CLNT_NAM_LN_2", concat(lit("LN2_"), col("id")))
    .withColumn("RLTD_CLNT_NAM_LN_1", concat(lit("LN1_"), col("id")))
    .withColumn("EXT_CUS_CLNT_ID", when(col("id") < 8, concat(lit("EXTCLNT_"), col("id"))).otherwise(lit("EXTCLNT_2")))
    .withColumn("AUDT_USER_ID", concat(lit("user"), col("id")))
    .withColumn("AUDT_PGM_ID", concat(lit("pgm"), col("id")))
    .withColumn("AUDT_TSTMP", current_timestamp())
    .withColumn("AUDT_INS_TSTMP", current_timestamp())
    .withColumn("AUDT_ACTN_CDE", lit("A"))
    .withColumn("AUD_CDD_ENTTYP", concat(lit("enttyp"), col("id")))
    .withColumn("AUD_CDD_TSTMP", current_timestamp())
    .withColumn("AUD_CDD_METHOD", concat(lit("method"), col("id")))
    .withColumn("DW_CLOUD_INSR_DTTM", current_timestamp())
    .withColumn("DW_CLOUD_UNIQID", uuid())
)
df_tcrrlcl.write.format("delta").mode("append").saveAsTable("dev_hub_bronze.lh_ax_idr.tcrrlcl")

# TCRROO1
df_tcrroo1 = (
    spark.range(0, 10)
    .withColumn("lh_actn_utc_dttm", current_timestamp())
    .withColumn("data_received_utc_dttm", current_timestamp())
    .withColumn("source_file_path", concat(lit("/path/file_"), col("id")))
    .withColumn("DRV_REQ_NBR", when(col("id") < 8, col("id")).otherwise(lit(3)))
    .withColumn("AX_CLNT_NBR", when(col("id") < 8, col("id") + 1000).otherwise(lit(1003)))
    .withColumn("SBRN_PGM_ID", when(col("id") < 8, lpad(col("id"), 2, "0")).otherwise(lit("03")))
    .withColumn("SBRN_PGM_ACNT_NBR", when(col("id") < 8, col("id") % 100).otherwise(lit(3)))
    .withColumn("SO_ID", when(col("id") < 8, col("id") % 10).otherwise(lit(3)))
    .withColumn("REQ_VERS_NBR", when(col("id") < 8, col("id") % 5).otherwise(lit(3)))
    .withColumn("CLNT_REQ_FUNC_TCDE", when(col("id") < 8, lpad(col("id"), 2, "0")).otherwise(lit("03")))
    .withColumn("EXT_CUS_CLNT_ID", when(col("id") < 8, concat(lit("extcus"), col("id"))).otherwise(lit("extcus3")))
    .withColumn("CLNT_SPLY_REQ_ID", concat(lit("supply"), col("id")))
    .withColumn("PAUTH_NBR", concat(lit("pauth"), col("id")))
    .withColumn("PRSH_GD_IND", when(col("id") % 2 == 0, lit("Y")).otherwise(lit("N")))
    .withColumn("DRIV_WAIT_IND", when(col("id") % 3 == 0, lit("Y")).otherwise(lit("N")))
    .withColumn("RGSTN_SEQ_NBR", lpad(col("id"), 4, "0"))
    .withColumn("INS_USER_ID", lit(None))
    .withColumn("INS_WLOC_CDE", lpad(col("id"), 4, "0"))
    .withColumn("CMNCTN_MD_CDE", lpad(col("id"), 2, "0"))
    .withColumn("CLNT_REQ_SUBM_TCDE", lpad(col("id"), 2, "0"))
    .withColumn("EDI_REQ_ID", col("id"))
    .withColumn("REQ_INIT_ENT_DT", date_add(current_date(), col("id")))
    .withColumn("AUDT_USER_ID", concat(lit("user"), col("id")))
    .withColumn("AUDT_PGM_ID", concat(lit("pgm"), col("id")))
    .withColumn("AUDT_TSTMP", current_timestamp())
    .withColumn("AUDT_INS_TSTMP", current_timestamp())
    .withColumn("AUDT_ACTN_CDE", lit("A"))
    .withColumn("DCMT_BTCH_ID", lpad(col("id"), 5, "0"))
    .withColumn("REQ_DB_LOC_CDE", lpad(col("id"), 2, "0"))
    .withColumn("REQ_DB_LOC_DT", date_add(current_date(), col("id") + 1))
    .withColumn("B3_TRANS_NBR", lpad(col("id"), 14, "0"))
    .withColumn("FOLLOWUP_IND", when(col("id") % 4 == 0, lit("Y")).otherwise(lit("N")))
    .withColumn("AUD_CDD_ENTTYP", concat(lit("enttyp"), col("id")))
    .withColumn("AUD_CDD_TSTMP", current_timestamp())
    .withColumn("AUD_CDD_METHOD", concat(lit("method"), col("id")))
    .withColumn("DW_CLOUD_INSR_DTTM", current_timestamp())
    .withColumn("DW_CLOUD_UNIQID", uuid())
)
df_tcrroo1.write.format("delta").mode("append").saveAsTable("dev_hub_bronze.lh_ax_idr.tcrroo1")

# TCRSTAT
df_tcrstat = (
    spark.range(0, 10)
    .withColumn("lh_actn_utc_dttm", current_timestamp())
    .withColumn("data_received_utc_dttm", current_timestamp())
    .withColumn("source_file_path", concat(lit("/path/file_"), col("id")))
    .withColumn("DRV_REQ_NBR", when(col("id") < 8, col("id")).otherwise(lit(4)))
    .withColumn("AX_CLNT_NBR", when(col("id") < 8, col("id") + 1000).otherwise(lit(1004)))
    .withColumn("SBRN_PGM_ID", when(col("id") < 8, lpad(col("id"), 2, "0")).otherwise(lit("04")))
    .withColumn("SBRN_PGM_ACNT_NBR", when(col("id") < 8, col("id") % 100).otherwise(lit(4)))
    .withColumn("SO_ID", when(col("id") < 8, col("id") % 10).otherwise(lit(4)))
    .withColumn("REQ_VERS_NBR", when(col("id") < 8, col("id") % 5).otherwise(lit(4)))
    .withColumn("CLNT_REQ_SCDE", when(col("id") < 8, lpad(col("id"), 2, "0")).otherwise(lit("04")))
    .withColumn("AUDT_INS_TSTMP", current_timestamp())
    .withColumn("STAT_XDT", date_add(current_date(), col("id")))
    .withColumn("STAT_EDT", date_add(current_date(), col("id") + 1))
    .withColumn("CLNT_REQ_STAT_TCDE", lpad(col("id"), 2, "0"))
    .withColumn("REQ_STAT_CAT_CDE", lpad(col("id"), 2, "0"))
    .withColumn("REQ_STAT_RCDE", lpad(col("id"), 2, "0"))
    .withColumn("OP_CELL_TCDE", lpad(col("id"), 2, "0"))
    .withColumn("AUDT_USER_ID", concat(lit("user"), col("id")))
    .withColumn("AUDT_PGM_ID", concat(lit("pgm"), col("id")))
    .withColumn("AUDT_TSTMP", current_timestamp())
    .withColumn("AUDT_ACTN_CDE", lit("A"))
    .withColumn("RCVR_IND", when(col("id") % 2 == 0, lit("Y")).otherwise(lit("N")))
    .withColumn("ORIG_REQ_INS_TSTMP", current_timestamp())
    .withColumn("INS_USER_ID", lit("0"))
    .withColumn("INS_WLOC_CDE", lpad(col("id"), 4, "0"))
    .withColumn("AUD_CDD_ENTTYP", concat(lit("enttyp"), col("id")))
    .withColumn("AUD_CDD_TSTMP", current_timestamp())
    .withColumn("AUD_CDD_METHOD", concat(lit("method"), col("id")))
    .withColumn("DW_CLOUD_INSR_DTTM", current_timestamp())
    .withColumn("DW_CLOUD_UNIQID", uuid())
    .withColumn("ELMT_ID_NBR", concat(lit("elmt"), col("id")))
    .withColumn("RSK_CRIT_ID", col("id"))
    .withColumn("EVNT_DTTM", current_timestamp())
)
df_tcrstat.write.format("delta").mode("append").saveAsTable("dev_hub_bronze.lh_ax_idr.tcrstat")
```
