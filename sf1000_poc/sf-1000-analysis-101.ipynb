##Block below will extract list of primary keys for each table in sf-1000


```python
# tables = [t.tableName for t in table_names]
tables = [row.tableName for row in tables_df.collect()]

keys_candidates = []
for t in tables:
    df = spark.table(f"samples.tpcds_sf1000.{t}")
    for c in df.columns:
        if c.endswith("_sk") or c.endswith("_id"):
            keys_candidates.append((t, c))

display(spark.createDataFrame(keys_candidates, ["table", "column"]))
```
